### 1. Loading the Input Datasets

```python
import pandas as pd

# Load the three individual CSV datasets
orders_df = pd.read_csv("Day9_Orders.csv")
customers_df = pd.read_csv("Day9_Customers.csv")
products_df = pd.read_csv("Day9_Products.csv")

print("Orders Shape:", orders_df.shape)
print("Customers Shape:", customers_df.shape)
print("Products Shape:", products_df.shape)

print("\nSample Orders:")
print(orders_df.head(2))
```

### 2. Merging Datasets using merge()

```python
# Merge orders with customers on Customer_ID, then merge with products on Product_ID
merged_df = orders_df.merge(customers_df, on='Customer_ID', how='left').merge(products_df, on='Product_ID', how='left')

print("Merged DataFrame Shape:", merged_df.shape)
print(merged_df.head(3))
```

### 3. Combining DataFrames using concat()

```python
# Demonstrate concatenation by splitting the merged dataset into two halves and joining them back
half1 = merged_df.iloc[:60]
half2 = merged_df.iloc[60:]

concatenated_df = pd.concat([half1, half2], ignore_index=True)
print("Concatenated DataFrame Shape:", concatenated_df.shape)
print("Sample Concatenated Data:")
print(concatenated_df[['Order_ID', 'Customer_ID', 'Product_ID']].head(2))
```

###4. Creating New Columns using apply()

```python
# Use apply() and a lambda function to calculate Total_Amount (Quantity * Unit_Price)
merged_df['Total_Amount'] = merged_df.apply(lambda row: row['Quantity'] * row['Unit_Price'], axis=1)

print("Sample with Calculated Total_Amount:")
print(merged_df[['Order_ID', 'Customer_Name', 'Product_Name', 'Quantity', 'Unit_Price', 'Total_Amount']].head(3))
```

###5. DateTime Operations and Extractions

```python
# Convert Order_Date to datetime and extract date components
merged_df['Order_Date'] = pd.to_datetime(merged_df['Order_Date'])

merged_df['Order_Year'] = merged_df['Order_Date'].dt.year
merged_df['Order_Month'] = merged_df['Order_Date'].dt.month_name()
merged_df['Order_Day_Name'] = merged_df['Order_Date'].dt.day_name()

print("Sample with Extracted Date Features:")
print(merged_df[['Order_ID', 'Order_Date', 'Order_Month', 'Order_Day_Name']].head(3))
```

###6. Exporting the Processed Dataset

```python
# Export the final processed dataframe to a CSV file
output_filename = "Processed_Ecommerce_Dataset.csv"
merged_df.to_csv(output_filename, index=False)

print(f"Final processed dataset successfully exported to '{output_filename}'!")
```
